```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_community.document_loaders import WebBaseLoader
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from IPython.display import display , Markdown

load_dotenv() 


import os

GROQ_API_KEY = os.getenv("GROQ_API_KEY")

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
urls = [
    "https://www.helwanchem.com.eg/%d9%83%d9%8a%d9%85%d8%a7%d9%88%d9%8a%d8%a7%d8%aa-%d8%a7%d9%84%d8%ac%d9%84%d9%81%d9%86%d8%a9-%d8%b9%d9%84%d9%89-%d8%a7%d9%84%d8%b3%d8%a7%d8%ae%d9%86/",
    "https://www.helwanchem.com.eg/%d9%83%d9%8a%d9%85%d8%a7%d9%88%d9%8a%d8%a7%d8%aa-%d9%85%d8%b9%d8%a7%d9%84%d8%ac%d8%a9-%d8%a7%d9%84%d8%ad%d8%af%d9%8a%d8%af/",
    "https://www.helwanchem.com.eg/%d9%83%d9%8a%d9%85%d8%a7%d9%88%d9%8a%d8%a7%d8%aa-%d9%85%d8%b9%d8%a7%d9%84%d8%ac%d8%a9-%d8%a7%d9%84%d8%a7%d9%84%d9%88%d9%85%d9%86%d9%8a%d9%88%d9%85/"
]

docs = [WebBaseLoader(url).load() for url in urls]
docs_list = [item for sublist in docs for item in sublist]


text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 250,
    chunk_overlap=20
)

doc_splits = text_splitter.split_documents(docs_list)

embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

vectorstore = InMemoryVectorStore.from_documents(
    documents=doc_splits,
    embedding=embeddings
)

retriever = vectorstore.as_retriever(k=5)
```

```python
from langsmith import traceable

@traceable()
def rag_bot(q:str)->dict :
    docs = retriever.invoke(q)
    doc_string = " ".join(doc.page_content for doc in docs)

    instructions = f"You are a helpful assistant.Respond to the user question based on following documents only: {doc_string}. if you didn't get the docs reply with i don't know."

    ai_msg = llm.invoke([
        {'role':'system','content':instructions},
        {'role':'user','content':q}
    ])

    return {'answer':ai_msg.content , "docs":docs}
```

```python
res = rag_bot("كيماويات الجلفنة على الساخن")
```

## Evaluator Metrics

```python
from typing_extensions import Annotated , TypedDict


class CorrectnessGrade(TypedDict):
    explanation : Annotated[str,"Explain your reasoning for the score"]
    correct:Annotated[bool,"True if the answer is correct,False otherwise"]

correctness_instruction = """Carefully analyze the input and provide a response that is factually correct, logically consistent, and precise. Avoid making assumptions beyond the given information. If the answer is uncertain, clearly state the uncertainty and do not guess. Ensure all calculations, code, or reasoning steps are accurate and verifiable.
Output Requirements:
Step-by-step reasoning if applicable.
Highlight any potential ambiguity or missing information.
Provide the final answer clearly and concisely."""
```

```python
grader_llm = llm.with_structured_output(CorrectnessGrade,method="json_schema",strict=True)
```

```python
def correctness(inputs:dict,outputs:dict,reference_outputs:dict)->bool:
    """An Evaluator for RAG"""
    answer = F"""\
Question:{inputs['question']}
Ground Truth Answer :{reference_outputs['answer']}
Student Answer:{outputs['answer']}"""
    
    grade = grader_llm.invoke([
        {'role':'system','content':correctness_instruction},
        {'role':'user','content':answer}
    ])

    return grade['correct']
```
